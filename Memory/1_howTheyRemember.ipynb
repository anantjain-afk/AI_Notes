## Problem with Short - Term Memory 

1. Short Term Memory is Fragile : It can break if code reset / server crashes , the messages array where we have stored all our conversation for this thread will get removed . 
That's why we use persistance : means we store messages using thread id in databases . 

2. the Context Window Problem : Long Conversation History (exceeds context) 
we solve this using some Techniques : like 
a. Trimming : only recent messages like recent 100 messages will be sent to llm only . 
b. summarization : keep a summarized version of the conversation (decresing the length of used context window )
c. Hybrid Context (Best) : Keep Summary too and Use Trimming too . so for trimmed chats , keep the summary of them , and don't create summary of ex. top 100 messages . 

Modern LLM use Hybrid Context . 

3. STM is thread - scoped : 
a. loss of user continuity across conversations 
b. learning never compounds over time 
c. Cross-thread reasoning is impossible 

Solution - We need a new king of memory 
Information that should survive : 
- beyond a single conversation and single session , sometimes even beyoun days or weeks 

Things like : 
who the user is 
how the system is expected to behave for them 
what tends to work and what ussually fails 
what dicision were already made in past .

- It has to be selective.
Only information that proves to be stable , userful , and reusable should survive beyong a single thread. Everything else should naturally fade away . 

### This is Long Term memory 


### Long-Term Memory 

Three types of Long-Term Memory : 
1. Episodic Memory - what happened 
- what it Stores :
Past events and experience s, usually tied to time or context . 
- why it exists : 
Answeres questions like : "What did we do last time?" , "have we tried this already"

2. Semantic Memory - what is true :
- What it stores 
Facts , knowledge , and stable information , that are not ried to a single event. 
- Why it exists : 
Answers question like : 'what do we know about this user?" , "What constraints should always apply?"

3. Procedural Memory - How to do things 
- What it Stores : 
Strategies,  rules and learned behaviours 
- why it exists : 
Answers questionos : 'what approach usually works?' ,'How should i behave in this situaltiion'
This memory compound overtime . 


## how does long term memory work ? 

- Step 1 - creation / update 
"Is anything from what just happened worth remembering beyong this conversation?"

what the system looks at : 
    User messages 
    Model Responses 
    Toop outcomes 

What happens in theis step 
The system: 
    extracts memory condidates 
    filters out noise 
    decides scope (user / agent / app)
    decides whether to: 
        create a new memory 
        update and existing one 
        ignore it 

- Step 2 - Storage 
'Where and how do we keep this memory?'

What storage actually means 
    Writing memory to a durable store 
    Assigning identifiers and metadata
    making it survive restarts and crashes 

Depending on the memory type this might be : 
    a relational database 
    a key-value store 
    a log 
    a vector database

- Step 3 - Retreival 
'given the current sitatuin , what should i remember right now ?'

What retrieval involvs 
The system:     
    looks at the current input 
    decides whether memory is needed 
    searches memory stores
    selects a small , relevant subset

Key point : retrieval is selective , not exhaustive .

- Step 4 - Injection 

'Hhow does memory actually infuence the model ? '
llm don't interact directly with LTM , we will retrieve the relevant memory from LTM and inject it STM and send it to the LLM

What injection really is 
    Retrieved memory is inserted into STM 
    it becomes part of the prompt 
    The model sees it as just more tokens 



- The Challenges : 
1. Deciding what is worth remembering 
2. Retreiving the right memory at the right time 
3. Orchestrating the entire system .
