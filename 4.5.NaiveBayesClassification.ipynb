Naive Bayes is a probabilistic machine learning algorithm mainly used for:

text classification
spam detection
sentiment analysis
document categorization

It is based on:

Bayes’ Theorem
probability

Naive Bayes predicts:

“Which class is most probable for this input?”

Example:

Spam or Not Spam
Positive or Negative
Fake or Real news

Why “Naive”?

Because it assumes:

all features are independent of each other  



it uses : bayes theorem : 
P(A∣B)=(P(B∣A)P(A)​)/P(B)
