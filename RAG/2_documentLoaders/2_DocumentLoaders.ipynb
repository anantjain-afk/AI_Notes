Document Loaders are LangChain utilities designed to extract data from various sources and convert it into a standardized format called a Document Object
. Every Document object contains two key properties:
page_content: The actual extracted text/data
.
metadata: Information about the data, such as its source, creation date, author name, or page number
.
Note: All Document Loaders return a list of Document objects
. Also, all standard community-built Document Loaders are found in the langchain_community.document_loaders package

```python
from langchain_community.document_loaders import TextLoader 
from langchain_community.document_loaders import PyPDFLoader
# from langchain_community.document_loaders import 

# TextLoader: 
# The simplest loader, used to read standard text files (.txt) and convert them into Document objects
loader = TextLoader('txt.txt' , encoding='utf-8')
docs = loader.load()

print(docs[0].page_content) # The actual text
print(docs[0].metadata) 
```

```python

# PyPDFLoader : 
# Used to load PDF files. Crucially, PyPDFLoader processes the file page by page,
#  meaning a 25-page PDF will return a list of exactly 25 Document objects, each with its own page_content and metadata (which tracks the specific page number)
# . It requires pypdf to be installed via pip
pdfloader = PyPDFLoader('Agentic_AI_Question_Bank.pdf')
docs = pdfloader.load()
# Returns the total number of pages loaded
print(len(docs)) 

# PyPDFLoader works best with text-based PDFs. If you are dealing with scanned images or heavy table structures in a PDF,
#  you should use alternative loaders like PDFPlumberLoader, UnstructuredPDFLoader, or AmazonTextractPDFLoader
print(docs[0].page_content)
print(docs[0].metadata)


```

```python
# C. DirectoryLoader
# Used when you have a folder containing multiple files (e.g., 100 PDFs) and you want to load them all at once
# . You provide the directory path, a glob pattern (to specify file types like *.pdf), and the underlying loader class to process those files
# but you can only use one loader class at a time. For example, if you have a directory with both PDFs and text files, you would need to run DirectoryLoader twice: once with PyPDFLoader for the PDFs and once with TextLoader for the text files.

from langchain_community.document_loaders import DirectoryLoader

pdf_loader = DirectoryLoader(
    './RAG/2_documentLoaders/Directory' ,
    glob="**/*.pdf",
    loader_cls=PyPDFLoader
)
txt_loader = DirectoryLoader(
    './RAG/2_documentLoaders/Directory' ,
    glob="**/*.txt",
    loader_cls=TextLoader
)
pdf_docs = pdf_loader.load()
txt_docs = txt_loader.load()

documents = pdf_docs + txt_docs
print(f"Loaded {len(documents)} documents")
```
