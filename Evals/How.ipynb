## LLM Application Eval Workflow : 

1. Define Task & Target  : Identify exactly what you are evaluating.

2. Define a Success Criteria  : Determine how you will measure success.

3. Build a Database Containing the correct examples . (Golden Dataset) : You need a baseline dataset to test the system against. This involves manually creating a dataset (typically 50 to 500 rows) mapping the input (the email content) to the expected output (the correct label)
It is highly recommended to build this dataset using actual past user data

4. Define an eval method : Decide who or what will calculate the accuracy by comparing the LLM's answers to the Golden Dataset. There are three main approaches: 

Automated (Python Code): Best for simple tasks like classification, where code can easily check if the generated label matches the expected label .

Human Evaluator: Highly accurate but slow and expensive 

LLM-as-a-Judge: If the application generates long textual responses (like a chatbot), simple code cannot compare paragraphs for semantic meaning. Instead, you can use another LLM to act as the evaluator

5. Run the Model : Pass your Golden Dataset through the application
. The system generates its predictions, and your evaluation method (e.g., Python code) calculates the final metric (e.g., an 80% accuracy score)


6. Analyze the resylts -> Improve the model -> go back to 5 and again try .: Investigate the errors (the 20% that failed) to find the scope for improvement 
you can improve : 
Tweaking the System Prompt to provide clearer instructions, 
Upgrading the LLM to a more capable model with more parameters

7. Iterate LLM evaluations are highly repeatable loops. After making improvements (like changing the prompt), you re-run the evaluation on the dataset
. If the score jumps to 90%, you might try upgrading the model to push it to 95%
. You iterate until you reach a satisfactory threshold for deployment

8. Deploy & Monitor -> Product Failures and create a feedback loop : 
Once deployed, you must continuously monitor the application in production
. If the system fails in the real world , you take that specific failed email and add it back into your Golden Dataset
. This enriches your dataset, allowing you to re-evaluate and improve the model continuously forever 


A very important takeaway is that one LLM-based application might have several different LLM evaluations running simultaneously 
a complex application like RAG might have :
One evaluation to test the Retriever's performance.
One evaluation to test the Embedding model.
One evaluation for the overall generated answer.
One evaluation to check system latency.
