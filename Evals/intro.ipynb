LLM evals are systematic , repeatable tests used to judge an LLM or LLM-powered system again a claer criteria. 

### Systematic 

it should not be random prompting 
instead of asking 5 questions casually and saying 'looking good' , we create planned datasets of test cases covering every edge cases for testing . 

### Repeatable 

the same eval should be runnable again . 
if we change the prompt , model, retrriver , or system instruction , we should be able to run the same test again and conquer results . 

This is how we know that has become better or worse . 

### Clear criteria 

We must define what good means .

Without criteria , we are only judges by vibes. 



### An Eval is not just a metric . An eval is the complete testing setup , it Includes : 

What are we evaluation ? 
What does good mean? 
What test cases are we using ? 
How are we judging the output  
when are we running it ? 
Which tool are we using ? 

The Goal is not just to get a good score . The goel is to answer practical questions like : 
Can the model be used for a particular task/application ? 
is this system good enough to ship ? 
Did prompot v2 improve over prompt v1 ?
is the RAG answer grounded in the retirved context ? 
is the agent completing the task correctly ? 
is the chatbot safe for real users ?
 

 


## Model Evals vs Application Evals 

LLM Evals encapsulates both Model Evals and Application Evals 

Disclamer : Model Evals or Application Evals are not official terms , these are just among developers . 

### Model Evals : model evals evaluates the model itself. the main idea is to evaluate the output of model .

Some important capabilities includes : 

1. Reasoning : Can it Reason ?
2. Knowledge : Can it answer knowledge- based questions ? 
3. Maths : Can it solve maths problems 
4. Coding : Can it write code 
5. instruction following : can it follow instructions . 
6. Long Context : Can it folow long documents 
7. Multilmodel understanding : Can it understand images . 
8. Tooluse : Can it use tools 

Model capabillities are generally evaluated using Benchmars . Some famous benchmarks include : 

General Knowledge + Reasoning : MMLU 
Maths : GSM8K 
Coding : HumanEval ,   SWE - bench 
Instruction Following : IFEval 
Long Context : Needle-in-a-Haystack 
Multi-Model : MMMU 

### Application Evals (Important):

Application Evals assess the behaviour and performance of an LLM-powered application , wheather at the level of the entire sustem or a specific component within it . 

In application eval we don't ask : Can the model do this ? 
instead application eval tells wheather the product works.
