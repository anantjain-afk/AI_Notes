Reason 2: Multiple Risk Categories
Beyond just getting the "correct" answer, LLM applications face various risks. An evaluation pipeline designed to test for accuracy cannot test for safety
. Risks are categorized into three main buckets
:
A. Application Quality (Does it do its job well?)
This category evaluates if the output is correct, relevant, complete, and follows instructions properly
.
RAG-Specific Quality: Context relevance, retriever recall, groundedness/faithfulness, and citation accuracy (ensuring the LLM cites the correct source document)
.
Agent-Specific Quality: Tool selection (did the agent pick the right tool for the job?), parameter correctness, overall task completion rate, and error recovery (can the agent fix itself if a tool fails?)
.
Multi-turn Chatbot Quality: Context retention (remembering previous chat history) and clarification behavior (asking the user for more details if a prompt is ambiguous)
.
B. Safety (Is the output harmful?)
Safety evaluations ensure the LLM does not generate dangerous or inappropriate text. Even if an answer is factually correct, it might be unsafe
. Key safety risks include:
Toxicity & Harmful Content: Preventing outputs related to self-harm, weapons, or illegal activities
.
Bias: Ensuring the bot doesn't discriminate based on a user's profile
.
Data Leaks (PII): Ensuring the bot doesn't accidentally reveal private information like credit card numbers or contact details
.
Prompt Injection / Jailbreaks: Testing if users can trick the LLM into ignoring its safety instructions
.
C. Operations (Is it fast, cheap, and reliable?)
Operational evaluations monitor the system's performance metrics when deployed
.
Key metrics include: Latency (time to generate an answer), Cost per request, Token efficiency, Error/Failure rates, and Latency under heavy user load
.
