An LLM eval methods is the mechanism you use to decide whether an LLM's output is good or not - the actual procedure that takes an output and produces a judgment about it . 
While an evaluation pipeline is the overarching process built to test an application, the Evaluation Method is the actual mechanism that carries out the execution and decides whether an LLM's output is good or bad


1. Programmatic / deterministic - code produces the verdict . 
the output is checked against a rule or a known correct answer by a program : exact match , regex , JSON schema validation , or running the output (eexecute the code , run the SQL ) and comparing results . 

fast , cheap , reproducable - but only works where correctness is mechanically checkable . 


2. Human - a person produces the verdict . 
people rate , rank or compare output directly , the most expensive and slowest , but the gold standard for nuanced , subjective , or high- stakes judgments . 

3. Model-graded(LLM-as-judge) - a model(LLM) produces the verdict . 

A strong LLM scores or anks the output against a rubric. Used where there's no single right answer but you can describe what 'good' look like - helpfulness , tone , groundness . 



2. Method 1: Programmatic Evaluation
This method uses hard-coded logic to evaluate an application. It is ideal for straightforward, mathematical, or matching tasks where nuance is not required

The Scenario: Evaluating the "Retriever" component of a RAG Chatbot to see if it fetches the correct documents from the vector database

The Metric (Recall@K): The success criteria is defined by Recall@K, which means: out of all the correct documents that actually exist in the database for a query, how many did the system successfully retrieve in its top K results? (e.g., If 2 documents contain the answer, and the retriever only finds 1, the recall is 50%
)
The Workflow:
Build a "Golden Dataset" of 50-100 questions where a human has manually mapped each question to the exact correct Document IDs
Pass the questions to the retriever programmatically
The code compares the retrieved Document IDs against the expected Document IDs in the Golden Dataset to calculate the recall for each question
Average the recall across all questions (e.g., getting a final score of 67%)

Next Steps: If the score is low, developers can improve it by changing the embedding model, expanding the query, increasing K, or adding a re-ranker

Why Programmatic? It is fast and cheap. You don't need to pay humans to simply match document IDs

3. Method 2: Human Evaluation
This method relies on human judgment. It is used when evaluating highly nuanced aspects of an application that code cannot measure

The Scenario: Evaluating an entire chat application for its Helpfulness (is the answer accurate, complete, and in the right tone?)

The Workflow:
Define a Rubric (e.g., a scale of 1 to 5, where 5 is perfectly helpful and 1 is completely unhelpful)
Build a Golden Dataset of just user questions (no answers provided) covering edge cases and normal queries
Feed the questions to the chatbot to generate answers
A human grader reads the question, reads the bot's answer, and assigns a score based on the rubric
Note on Multiple Graders: Often, multiple humans grade the same answers. If their scores constantly disagree, it indicates the rubric is ambiguous and needs to be rewritten

Other Types of Human Evals: Besides direct grading, humans also do Red Teaming (intentionally attacking the system to find breaking points), A/B Testing in production, creating Golden Datasets, and acting as a "Human-in-the-loop" for complex edge cases
Pros & Cons: Human judgment is highly reliable
 However, the major disadvantage is cost. You cannot scale human evaluations to test thousands of queries quickly

4. Method 3: Model-Graded Evaluation (LLM-as-a-Judge)
When programmatic evaluation is impossible due to the nuance of the task, and human evaluation is too expensive to scale, the industry standard is to use LLM-as-a-Judge

The Scenario: Building a platform to automatically grade subjective written answers for UPSC (Civil Services) exams

The Target & Success Criteria: The system is deemed successful if the LLM platform assigns the exact same marks to a student's answer as a real human expert would

The Workflow:
Define a detailed rubric for a specific exam question (e.g., "Award marks if the answer mentions ethical governance, provides examples, and has a balanced conclusion")
Have a human expert evaluate 50-100 real student answers using this rubric to create the Golden Dataset
Pass the student's answer, the question, and the exact rubric into an evaluator LLM via a carefully crafted system prompt
The LLM acts as the judge, returning the marks it believes the student deserves along with a justification
The Metric (Mean Absolute Error - MAE): To evaluate if the system works, calculate the Mean Absolute Error between the Human's marks and the LLM's marks
If the MAE is 2.3, it means the LLM deviates from the human by an average of 2.3 marks
The goal of the developer is to iterate on the system prompt or upgrade the model to drive the MAE as close to 0 as possible

5. Crucial Concept: Reference-Based vs. Reference-Free Evaluations
LLM evaluations can be divided into these two categories based on how the Golden Dataset is structured:
Reference-Based Evaluations:
What it is: The correct answer (or the reference) is known and written down in advance in the Golden Dataset. You grade the system by comparing its output directly against this reference

Examples: The Programmatic Retriever eval (we already knew the correct Document IDs)
and the UPSC grading eval (we already knew the correct marks assigned by the human)

Reference-Free Evaluations:
What it is: There is no predefined "correct" answer in the dataset. The output is judged purely on its own quality against a set standard or rubric

Example: The Human evaluation of Chatbot Helpfulness. The dataset only contained questions. The human judged the generated answer directly against a 1-to-5 rubric scale without having a "perfect reference answer" to compare it to
