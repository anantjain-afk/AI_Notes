## Observability 
 
it is the ability to understand a system's internal working or states by examining the external outputs , traces , logs , metrics . 
It allows you to diagnose issues , understand performace , improve reliability by analyzing data generated by the system . 

because of it , we are able to understand the 'why' of something is happening even if you don't anticipate the problem . 


## Langsmith 

Langsmith is a unified observability and evaluation platform where teams can debug , test , and montitor an AI app performance 

### What Does langsmith trace 

input and output 
all intermediate steps 
latency 
tokens used 
costs
errors 
tags 
metadata 



there are three core concepts of langsmith : 
1. Porject 
2. Trace 
3. Run 

let's unterstand these concepts using a simple example :
we have a simple llm app where user will come to the app , give prompt and that prompt will be given to the llm and the output the llm will be shown to the user using a parser 

user -> prompt -> llm -> parser 

this whole setting is one project 
now , one complete execution of this project is known as Trace in Langsmith . 
and as this proejct has multiple components , one execution of any component is one Run  .
