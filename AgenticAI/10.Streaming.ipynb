in LLM , streaming means the model start returning token as soon as they're generated , instead of waiting for the entire response to be generated . 

Why Streaming ? 
1. Faster Response Time - Low Drop-off rates 
2. Mimics human behaviour like conversation . (it builds trusts , feel alive , and keeps the user engaged )
3. Important for Multi-tool UIs (Like Alexa , its very much needed)
4. Better UX for long output such as code .
5. You can cancel midway , and save tokens .
6. You can interleave Ui updates , e.g., show 'Thinking...' show tools used.
